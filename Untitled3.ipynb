```python
import torch
import torch.nn as nn
import torch.optim as optim

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

sentence = "the dog is barking"
words = sentence.split()

vocab = list(set(words))
word_to_idx = {w: i for i, w in enumerate(vocab)}
idx_to_word = {i: w for w, i in word_to_idx.items()}
vocab_size = len(vocab)

input_words = words[:3]
target_word = words[3]

input_ids = [word_to_idx[w] for w in input_words]
target_id = word_to_idx[target_word]

input_tensor = torch.tensor([input_ids], dtype=torch.long).to(device)
target_tensor = torch.tensor([target_id], dtype=torch.long).to(device)

class MyRNN(nn.Module):
    def __init__(self, vocab_size, emb_dim, hidden_dim):
        super(MyRNN, self).__init__()
        self.embed = nn.Embedding(vocab_size, emb_dim)
        self.rnn = nn.RNN(emb_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, x):
        x = self.embed(x)
        h0 = torch.zeros(1, x.size(0), hidden_dim).to(device)
        out, _ = self.rnn(x, h0)
        out = self.fc(out[:, -1, :])
        return out

emb_dim = 8
hidden_dim = 20
epochs = 50
lr = 0.005

model = MyRNN(vocab_size, emb_dim, hidden_dim).to(device)
loss_fn = nn.CrossEntropyLoss()
opt = optim.Adam(model.parameters(), lr=lr)

for epoch in range(epochs):
    model.train()
    out = model(input_tensor)
    loss = loss_fn(out, target_tensor)

    opt.zero_grad()
    loss.backward()
    opt.step()

    if (epoch + 1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}')

with torch.no_grad():
    pred = model(input_tensor)
    pred_idx = torch.argmax(pred, dim=1).item()
    pred_word = idx_to_word[pred_idx]
    print(f"Predicted word: {pred_word}")
```
